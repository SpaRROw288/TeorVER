```python
cards = ['А', 'А', 'А', 'Н', 'Н', 'С']

all_combin = len(list(permutations(cards)))

total_pineapple = 0
for el in list(permutations(cards)):
    word = ''.join(el)
    if word == 'АНАНАС':
        total_pineapple += 1
print(total_pineapple / all_combin)

```

```python
from itertools import permutations

var = ['M','M','M','M','M',         'W','W','W', 'W','W']

all_var_sit = len(list(permutations(var)))

k=0 
for el in permutations(var):
    ok = True
    for i in range(0, len(el)):
        if el[i] == el[(i+1) % len(el)]:
            ok = False
            break
    if ok:
        k+=1

print(k)
print(k / all_var_sit)
```

```python


n = sc.comb(36, 7, exact=True)

# а) 7 пассажиров попали в 2 купе
m1 = sc.comb(9, 2, exact=True) * sc.comb(8, 7, exact=True)
p_a = m1 / n

# б) 7 пассажиров попали в 7 купе 
m2 = sc.comb(9, 7, exact=True) * 4**7
p_b = m2 / n

# в) 7 пассажиров попали в 3 купе ( 4+2+1, 3+3+1, 3+2+2)
C4 = lambda k: sc.comb(4, k, exact=True)
inner = 6*C4(4)*C4(2)*C4(1) + 3*C4(3)*C4(3)*C4(1) + 3*C4(3)*C4(2)*C4(2)
m3 = sc.comb(9, 3, exact=True) * inner
p_d = m3 / n

print(f"P(A) = {p_a:.7f}")
print(f"P(B) = {p_b:.5f}")
print(f"P(D) = {p_d:.5f}")
```
